# 1. Business Understanding

### 1.1 Pernyataan Masalah
Pencemaran udara merupakan salah satu masalah lingkungan yang dapat berdampak pada kesehatan masyarakat. Salah satu polutan yang perlu diperhatikan adalah karbon monoksida (CO), yaitu gas yang dihasilkan dari pembakaran tidak sempurna dan dapat mengganggu kemampuan darah dalam mengangkut oksigen.

Pengukuran konsentrasi CO secara akurat membutuhkan alat analyzer referensi yang memerlukan biaya, kalibrasi, dan perawatan. Sementara itu, sensor kimia berbiaya rendah lebih mudah diterapkan dalam jumlah banyak, tetapi hasil pengukurannya dapat dipengaruhi oleh suhu, kelembapan, gas lain, dan perubahan kinerja sensor.

Permasalahan yang diangkat dalam proyek ini adalah bagaimana memanfaatkan data respons sensor dan kondisi lingkungan untuk memperkirakan konsentrasi CO secara lebih akurat. Untuk menjawab permasalahan tersebut, diterapkan machine learning dengan pendekatan supervised learning berupa regresi.

Model akan dilatih menggunakan data historis kualitas udara dari dataset Air Quality dan dievaluasi menggunakan MAE, RMSE, dan R² pada data uji yang tidak digunakan selama pelatihan. Hasilnya diharapkan dapat menunjukkan kemampuan model dalam memperkirakan konsentrasi CO serta memberikan informasi mengenai variabel yang berkontribusi terhadap hasil prediksi.

### 1.2 Business Objective
Tujuan bisnis proyek ini adalah mengembangkan pendekatan prediksi konsentrasi CO yang dapat membantu pemantauan kualitas udara dengan memanfaatkan data sensor dan variabel lingkungan yang tersedia.

Secara khusus, proyek ini bertujuan untuk:
1. Menghasilkan estimasi konsentrasi CO berdasarkan data respons sensor dan kondisi lingkungan.
2. Membandingkan beberapa model regresi untuk mengetahui model yang memberikan performa prediksi terbaik pada data uji.
3. Mengidentifikasi variabel yang paling informatif dalam memprediksi konsentrasi CO.
4. Memberikan informasi yang dapat menjadi bahan pertimbangan bagi pengelola kualitas udara, pengembang sensor, serta peneliti dalam pengembangan sistem pemantauan kualitas udara.

Hasil proyek ini merupakan evaluasi model berbasis data historis, sehingga belum dapat langsung dianggap sebagai sistem peringatan dini yang siap digunakan tanpa pengujian lebih lanjut.

### 1.3 Data Science Objective
Proyek ini bertujuan membangun model supervised learning untuk memprediksi konsentrasi karbon monoksida (CO(GT)) menggunakan data kualitas udara dari dataset Air Quality.

Variabel target yang digunakan adalah CO(GT), sedangkan variabel prediktor dipilih dari respons sensor dan kondisi lingkungan yang relevan dengan skenario penggunaan model.

Model yang akan dibandingkan meliputi:
1. Linear Regression sebagai model baseline.
2. Random Forest Regressor untuk mempelajari hubungan nonlinier dan interaksi antarvariabel.

Evaluasi dilakukan menggunakan MAE, RMSE, dan R². Data latih dan data uji akan dipisahkan berdasarkan urutan waktu untuk mengurangi risiko kebocoran informasi dari periode masa depan ke periode pelatihan. Analisis feature importance juga dilakukan untuk mengetahui variabel yang informatif bagi model terbaik.

### 1.4 Pertanyaan Penelitian
Berdasarkan permasalahan dan tujuan proyek, pertanyaan penelitian yang dirumuskan adalah:
1. Seberapa akurat model regresi dalam memprediksi konsentrasi karbon monoksida (CO) berdasarkan respons sensor kualitas udara dan kondisi lingkungan pada dataset Air Quality?
2. Di antara algoritma Linear Regression, dan Random Forest model manakah yang menghasilkan kinerja prediksi terbaik berdasarkan nilai MAE, RMSE, dan R² pada data pengujian?
3. Variabel sensor dan kondisi lingkungan apa yang paling berkontribusi terhadap hasil prediksi konsentrasi CO berdasarkan analisis kepentingan fitur (feature importance)?

### 1.5 Metriks Keberhasilan
Keberhasilan proyek dinilai berdasarkan performa prediksi model dan kemampuan analisis menghasilkan informasi yang dapat diinterpretasikan.

Metrik evaluasi yang digunakan adalah:
1. Mean Absolute Error (MAE): mengukur rata-rata selisih absolut antara nilai CO aktual dan hasil prediksi. Nilai yang lebih rendah menunjukkan kesalahan prediksi rata-rata yang lebih kecil.
2. Root Mean Squared Error (RMSE): mengukur besarnya kesalahan prediksi dengan memberikan penalti lebih besar pada kesalahan yang besar. Nilai yang lebih rendah menunjukkan performa yang lebih baik.
3. Coefficient of Determination (R²): mengukur kemampuan model dalam menjelaskan variasi nilai CO dibandingkan dengan baseline prediksi rata-rata. Nilai yang lebih tinggi umumnya menunjukkan performa yang lebih baik.

Target awal proyek adalah memperoleh R² paling mendekati 0,85 pada data uji serta menghasilkan RMSE dan MAE yang lebih rendah daripada model baseline.

### 3.3 Jumlah pemeriksaan nilai kosong dan duplikat

```python
# Menghitung nilai kosong yang terbaca sebagai NaN
jumlah_nan = df.isna().sum()
jumlah_duplicated = df.duplicated().sum()

print("Jumlah nilai kosong (NaN) per kolom:")
print(jumlah_nan[jumlah_nan > 0].sort_values(ascending=False))

print()
print("Jumlah nilai duplikat:")
print(jumlah_duplicated)

# Menghitung outlier menggunakan metode IQR
kolom_numerik = df.select_dtypes(include='number').columns

hasil_outlier = []

for kolom in kolom_numerik:
    Q1 = df[kolom].quantile(0.25)
    Q3 = df[kolom].quantile(0.75)
    IQR = Q3 - Q1

    batas_bawah = Q1 - 1.5 * IQR
    batas_atas = Q3 + 1.5 * IQR

    jumlah_outlier = (
        (df[kolom] < batas_bawah) |
        (df[kolom] > batas_atas)
    ).sum()

    hasil_outlier.append({
        'Kolom': kolom,
        'Jumlah Outlier': jumlah_outlier,
        'Batas Bawah': batas_bawah,
        'Batas Atas': batas_atas
    })

hasil_outlier = pd.DataFrame(hasil_outlier)

display("Jumlah outlier per kolom:")
print(hasil_outlier.to_string(index=False))
```

**Penjelasan**: Terdapat 114 nilai kosong pada setiap kolom utama, sedangkan kolom Unnamed: 15 dan Unnamed: 16 masing-masing memiliki 9.471 nilai kosong karena seluruh kolom tersebut tidak berisi data. Selain itu, ditemukan 113 baris duplikat yang memiliki isi sama dengan baris lainnya. Temuan ini menunjukkan bahwa dataset masih mengandung nilai kosong, kolom yang tidak informatif, dan data duplikat. Namun, pada tahap EDA ini, data tetap dipertahankan dalam kondisi aslinya. ### 4.3 Persiapan Data
Melanjutkan `df_clean` dari cleaning 4.2: kolom kosong (`Unnamed: 15` dan `Unnamed: 16`) dibuang, `Date` dan `Time` digabung menjadi `Datetime`, lalu data diurutkan berdasarkan waktu.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor

data = df_clean.drop(columns=["Unnamed: 15", "Unnamed: 16"]).copy()

data["Datetime"] = pd.to_datetime(
    data["Date"] + " " + data["Time"].str.replace(".", ":", regex=False),
    format="%d/%m/%Y %H:%M:%S"
)
data = data.drop(columns=["Date", "Time"]).sort_values("Datetime").reset_index(drop=True)

# Fitur waktu (data berupa deret waktu per jam):
# jam dibuat siklik (sin/cos) agar jam 23 dan jam 0 dianggap berdekatan
data["jam_sin"] = np.sin(2 * np.pi * data["Datetime"].dt.hour / 24)
data["jam_cos"] = np.cos(2 * np.pi * data["Datetime"].dt.hour / 24)
data["akhir_pekan"] = (data["Datetime"].dt.day_of_week >= 5).astype(int)

print("Ukuran data:", data.shape)
print("Total nilai kosong:", int(data.isna().sum().sum()))
print("Rentang waktu:", data["Datetime"].min(), "sampai", data["Datetime"].max())
```

**Penjelasan**: Data berukuran 9.357 baris dan 17 kolom (14 kolom awal ditambah 3 fitur waktu: `jam_sin`, `jam_cos`, dan `akhir_pekan`), tanpa nilai kosong, dan mencakup 10 Maret 2004 sampai 4 April 2005. Jam dibuat siklik (sin/cos) agar jam 23 dan jam 0 dianggap berdekatan, sehingga sifat deret waktu data tetap dimanfaatkan oleh Linear Regression dan Random Forest.

### 4.4 Penanganan Data Hilang pada Target dan Pembagian Data Latih dan Uji
Nilai `-200` adalah penanda data hilang. Pada target `CO(GT)`, nilai tersebut **tidak diisi dengan rata-rata** karena nilai aslinya tidak diketahui; model akan belajar dan dinilai terhadap angka karangan sehingga R² tertahan rendah. Baris dengan target hilang dibuang. Baris saat seluruh sensor hilang bersamaan (alat mati) juga dibuang karena tidak ada masukan yang dapat dipakai model.

Setelah itu data dibagi secara kronologis: 80% data paling awal untuk latih dan 20% paling akhir untuk uji, agar informasi dari masa depan tidak bocor ke data latih.

```python
from sklearn.model_selection import train_test_split

target = "CO(GT)"

fitur_sensor_env = [
    "PT08.S1(CO)", "PT08.S2(NMHC)", "PT08.S3(NOx)",
    "PT08.S4(NO2)", "PT08.S5(O3)", "T", "RH", "AH"
]

fitur_waktu = ["jam_sin", "jam_cos", "akhir_pekan"]

fitur_referensi = [
    "NMHC(GT)", "C6H6(GT)", "NOx(GT)", "NO2(GT)"
]

fitur_model = fitur_sensor_env + fitur_waktu
kandidat = fitur_model + fitur_referensi

# -200 adalah penanda data hilang -> ubah menjadi NaN
kolom_ukur = [target] + fitur_sensor_env + fitur_referensi
data_bersih = data.copy()
data_bersih[kolom_ukur] = data_bersih[kolom_ukur].replace(-200, np.nan)

n_awal = len(data_bersih)

# 1) Baris yang target CO(GT)-nya hilang dibuang (nilai aslinya tidak diketahui,
#    sehingga tidak boleh diisi mean karena akan menjadi "jawaban palsu")
data_bersih = data_bersih.dropna(subset=[target])
n_target = n_awal - len(data_bersih)

# 2) Baris saat seluruh sensor/lingkungan hilang bersamaan (alat mati) dibuang
sebelum = len(data_bersih)
data_bersih = data_bersih.dropna(subset=fitur_sensor_env, how="all")
n_alat_mati = sebelum - len(data_bersih)

data_bersih = data_bersih.reset_index(drop=True)

print("Baris awal                     :", n_awal)
print("Dibuang (target CO hilang)     :", n_target)
print("Dibuang (semua sensor hilang)  :", n_alat_mati)
print("Baris tersisa                  :", len(data_bersih))

X = data_bersih[kandidat]
y = data_bersih[target]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    shuffle=False
)

idx_batas = len(X_train)
print("\nShape Data Train dan Test:")
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
print("\nPeriode train:", data_bersih["Datetime"].iloc[0], "sampai", data_bersih["Datetime"].iloc[idx_batas - 1])
print("Periode test :", data_bersih["Datetime"].iloc[idx_batas], "sampai", data_bersih["Datetime"].iloc[-1])
```

**Penjelasan**: Dari 9.357 baris, 1.683 baris dibuang karena target `CO(GT)` hilang dan 330 baris dibuang karena seluruh sensor hilang bersamaan, sehingga tersisa 7.344 baris. Data latih berjumlah 5.875 baris (10 Maret 2004 sampai 28 Januari 2005) dan data uji 1.469 baris (28 Januari sampai 4 April 2005). Karena pembagian kronologis, model diuji pada periode yang belum pernah dilihat saat pelatihan. Model hanya dapat dipakai saat sensor aktif.

### 4.5 Tangani -200 pada Fitur dengan Mean
Target sudah bersih dari `-200` pada tahap 4.4, sehingga imputasi mean hanya dipakai untuk `-200` yang tersisa di fitur. Nilai mean dihitung dari data latih saja.

```python
# -200 pada target sudah dibuang di 4.4, sehingga imputasi mean hanya
# dipakai untuk -200 yang tersisa di fitur (mean dihitung dari data latih saja)
X_train = X_train.replace(-200, np.nan)
X_test = X_test.replace(-200, np.nan)

mean_X = X_train.mean()

X_train = X_train.fillna(mean_X)
X_test = X_test.fillna(mean_X)

print("Shape setelah imputasi:")
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)

print("\nTotal missing value:")
print("X_train:", X_train.isna().sum().sum())
print("X_test :", X_test.isna().sum().sum())
print("y_train:", y_train.isna().sum())
print("y_test :", y_test.isna().sum())
```

# 5. Modelling

### 5.1 Modeling Linear Regression

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import GridSearchCV, TimeSeriesSplit

# Model Linear Regression
model_lr = LinearRegression()

# Hyperparameter yang diuji
param_grid_lr = {
    "fit_intercept": [True, False]
}

grid_lr = GridSearchCV(
    estimator=model_lr,
    param_grid=param_grid_lr,
    scoring="r2",
    n_jobs=-1
)

grid_lr.fit(X_train_linear, y_train)

print("Best Parameter:", grid_lr.best_params_)
print("Best R²:", grid_lr.best_score_)

model_linear = grid_lr.best_estimator_

print("Linear Regression selesai dilatih.")
```

### 5.2 Modeling dengan RandomForest

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import RandomizedSearchCV, TimeSeriesSplit
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

# Hyperparameter yang diuji
param_distributions_rf = {
    "n_estimators": [100, 200, 300, 400, 500],
    "max_depth": [None, 10, 20, 30, 50],
    "min_samples_split": [2, 5, 10, 15, 20],
    "min_samples_leaf": [1, 2, 4, 6, 10],
    "max_features": ["sqrt", "log2", 0.5, 0.7, 1.0],
    "bootstrap": [True, False],
    "criterion": [
        "squared_error",
        "absolute_error",
        "poisson"
    ],
    "max_leaf_nodes": [None, 20, 40, 60, 100],
    "min_impurity_decrease": [0.0, 0.0001, 0.001, 0.01, 0.1],
    "ccp_alpha": [0.0, 0.0001, 0.001, 0.01, 0.1]
}

# Validasi silang berdasarkan urutan waktu
tscv = TimeSeriesSplit(n_splits=5)

# Randomized Search
random_search_rf = RandomizedSearchCV(
    estimator=RandomForestRegressor(
        random_state=42,
        n_jobs=1
    ),
    param_distributions=param_distributions_rf,
    n_iter=50,
    cv=tscv,
    scoring="r2",
    random_state=42,
    n_jobs=1,
    verbose=1,
    error_score="raise"
)

try:
    # Training
    random_search_rf.fit(X_train_rf, y_train)

    # Hyperparameter terbaik
    best_params = random_search_rf.best_params_

    print("Best Parameter:", best_params)
    print("Best R² CV:", random_search_rf.best_score_)

    # Model terbaik
    model_rf = random_search_rf.best_estimator_

    # Prediksi data testing
    y_pred_rf = model_rf.predict(X_test_rf)

    # Evaluasi
    rmse_rf = np.sqrt(mean_squared_error(y_test, y_pred_rf))
    mae_rf = mean_absolute_error(y_test, y_pred_rf)
    r2_rf = r2_score(y_test, y_pred_rf)

    print("\nEvaluasi Random Forest pada Data Testing")
    print("RMSE:", rmse_rf)
    print("MAE:", mae_rf)
    print("R²:", r2_rf)

except Exception as e:
    print("Jenis Error:", type(e).__name__)
    print("Pesan Error:", e)
```

```python
model_rf.fit(X_train_rf, y_train)

print("Random Forest selesai dilatih.")
```
