# 1. Business Understanding

### 1.1 Pernyataan Masalah
Pencemaran udara merupakan salah satu masalah lingkungan yang dapat berdampak pada kesehatan masyarakat. Salah satu polutan yang perlu diperhatikan adalah karbon monoksida (CO), yaitu gas yang dihasilkan dari pembakaran tidak sempurna dan dapat mengganggu kemampuan darah dalam mengangkut oksigen.

Pengukuran konsentrasi CO secara akurat membutuhkan alat analyzer referensi yang memerlukan biaya, kalibrasi, dan perawatan. Sementara itu, sensor kimia berbiaya rendah lebih mudah diterapkan dalam jumlah banyak, tetapi hasil pengukurannya dapat dipengaruhi oleh suhu, kelembapan, gas lain, dan perubahan kinerja sensor.

Permasalahan yang diangkat dalam proyek ini adalah bagaimana memanfaatkan data respons sensor dan kondisi lingkungan untuk memperkirakan konsentrasi CO secara lebih akurat. Untuk menjawab permasalahan tersebut, diterapkan machine learning dengan pendekatan supervised learning berupa regresi.

Model akan dilatih menggunakan data historis kualitas udara dari dataset Air Quality dan dievaluasi menggunakan MAE, RMSE, dan R² pada data uji yang tidak digunakan selama pelatihan. Hasilnya diharapkan dapat menunjukkan kemampuan model dalam memperkirakan konsentrasi CO serta memberikan informasi mengenai variabel yang berkontribusi terhadap hasil prediksi.

### 1.2 Business Objective
Tujuan bisnis proyek ini adalah mengembangkan pendekatan prediksi konsentrasi CO yang dapat membantu pemantauan kualitas udara dengan memanfaatkan data sensor dan variabel lingkungan yang tersedia.

Secara khusus, proyek ini bertujuan untuk:
1. Menghasilkan estimasi konsentrasi CO berdasarkan data respons sensor dan kondisi lingkungan.
2. Membandingkan beberapa model regresi untuk mengetahui model yang memberikan performa prediksi terbaik pada data uji.
3. Mengidentifikasi variabel yang paling informatif dalam memprediksi konsentrasi CO.
4. Memberikan informasi yang dapat menjadi bahan pertimbangan bagi pengelola kualitas udara, pengembang sensor, serta peneliti dalam pengembangan sistem pemantauan kualitas udara.

Hasil proyek ini merupakan evaluasi model berbasis data historis, sehingga belum dapat langsung dianggap sebagai sistem peringatan dini yang siap digunakan tanpa pengujian lebih lanjut.

### 1.3 Data Science Objective
Proyek ini bertujuan membangun model supervised learning untuk memprediksi konsentrasi karbon monoksida (CO(GT)) menggunakan data kualitas udara dari dataset Air Quality.

Variabel target yang digunakan adalah CO(GT), sedangkan variabel prediktor dipilih dari respons sensor dan kondisi lingkungan yang relevan dengan skenario penggunaan model.

Model yang akan dibandingkan meliputi:
1. Linear Regression sebagai model baseline.
2. Random Forest Regressor untuk mempelajari hubungan nonlinier dan interaksi antarvariabel.

Evaluasi dilakukan menggunakan MAE, RMSE, dan R². Data latih dan data uji akan dipisahkan berdasarkan urutan waktu untuk mengurangi risiko kebocoran informasi dari periode masa depan ke periode pelatihan. Analisis feature importance juga dilakukan untuk mengetahui variabel yang informatif bagi model terbaik.

### 1.4 Pertanyaan Penelitian
Berdasarkan permasalahan dan tujuan proyek, pertanyaan penelitian yang dirumuskan adalah:
1. Seberapa akurat model regresi dalam memprediksi konsentrasi karbon monoksida (CO) berdasarkan respons sensor kualitas udara dan kondisi lingkungan pada dataset Air Quality?
2. Di antara algoritma Linear Regression, dan Random Forest model manakah yang menghasilkan kinerja prediksi terbaik berdasarkan nilai MAE, RMSE, dan R² pada data pengujian?
3. Variabel sensor dan kondisi lingkungan apa yang paling berkontribusi terhadap hasil prediksi konsentrasi CO berdasarkan analisis kepentingan fitur (feature importance)?

### 1.5 Metriks Keberhasilan
Keberhasilan proyek dinilai berdasarkan performa prediksi model dan kemampuan analisis menghasilkan informasi yang dapat diinterpretasikan.

Metrik evaluasi yang digunakan adalah:
1. Mean Absolute Error (MAE): mengukur rata-rata selisih absolut antara nilai CO aktual dan hasil prediksi. Nilai yang lebih rendah menunjukkan kesalahan prediksi rata-rata yang lebih kecil.
2. Root Mean Squared Error (RMSE): mengukur besarnya kesalahan prediksi dengan memberikan penalti lebih besar pada kesalahan yang besar. Nilai yang lebih rendah menunjukkan performa yang lebih baik.
3. Coefficient of Determination (R²): mengukur kemampuan model dalam menjelaskan variasi nilai CO dibandingkan dengan baseline prediksi rata-rata. Nilai yang lebih tinggi umumnya menunjukkan performa yang lebih baik.

Target awal proyek adalah memperoleh R² minimal 0,85 pada data uji serta menghasilkan RMSE dan MAE yang lebih rendah daripada model baseline.

# 4. Data Cleaning and Data Preprocessing

### 4.1 Menghapus baris dan kolom yang tidak berisi informasi serta hapus data duplikat

```python
print("Ukuran data:", df.shape)
print("\nJumlah nilai kosong:")
print(df.isna().sum())
print("\nJumlah baris duplikat:", df.duplicated().sum())
```

```python
df_clean = df.copy()
df_clean = df_clean.dropna(axis=0, how="all")
df_clean = df_clean.drop_duplicates()

print("Ukuran setelah menghapus baris dan kolom kosong:", df_clean.shape)
```

```python
print("Ukuran data:", df_clean.shape)
print("\nJumlah nilai kosong:")
print(df_clean.isna().sum())
print("\nJumlah baris duplikat:", df_clean.duplicated().sum())
```

### 4.2 Mengubah tipe data dan menangani nilai -200

```python
kolom_numerik = [
    "CO(GT)", "PT08.S1(CO)", "NMHC(GT)", "C6H6(GT)",
    "PT08.S2(NMHC)", "NOx(GT)", "PT08.S3(NOx)",
    "NO2(GT)", "PT08.S4(NO2)", "PT08.S5(O3)",
    "T", "RH", "AH"
]

kolom_numerik = [
    kol for kol in kolom_numerik if kol in df_clean.columns
]

jumlah_minus_200 = (df_clean[kolom_numerik] == -200).sum()

print("Jumlah nilai -200 pada setiap fitur:")
display(jumlah_minus_200.to_frame(name="Jumlah -200"))

# Mengubah kolom menjadi numerik terlebih dahulu
for kol in kolom_numerik:
    df_clean[kol] = pd.to_numeric(
        df_clean[kol].astype(str).str.replace(",", ".", regex=False),
        errors="coerce"
    )

# Menangani -200 dengan mean setiap fitur
for kol in kolom_numerik:
    mean_kolom = df_clean.loc[df_clean[kol] != -200, kol].mean()
    df_clean[kol] = df_clean[kol].replace(-200, mean_kolom)

# Menampilkan jumlah -200 setelah penanganan
print("Jumlah nilai -200 setelah penanganan:")
display((df_clean[kolom_numerik] == -200).sum().to_frame(name="Jumlah -200 Sesudah"))
```
